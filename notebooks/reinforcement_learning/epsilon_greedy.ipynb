```python
import numpy as np
from scipy import stats
from matplotlib import pyplot as plt
```

```python
k = 4

R_mean = [*range(1, k+1)]
R_sd = 1
```

```python
plt.figure("Reward Distributions")

x_axis = np.linspace(-2, k+3, 1000)

R_pdf = np.array([stats.norm.pdf(x_axis, loc=mean, scale=R_sd) for mean in R_mean])

for pdf in R_pdf:
    plt.plot(x_axis, pdf)

for mean, pdf in zip(R_mean, R_pdf):
    plt.plot([mean, mean], [0, np.max(pdf)], '--k')

plt.xlabel("Reward (R)")
plt.ylabel("Probability")

plt.show()
```
